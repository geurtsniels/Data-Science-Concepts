#### Slicing
In addition to accessing list elements one at a time, Python provides concise syntax to access sublists; this is known as slicing. Remember Python counts from 0.

```python
nums = list(range(5))    # range is a built-in function that creates a list of integers
print(nums)
print(nums[2:4])    # Get a slice from index 2 to 4 (exclusive)
print(nums[2:])     # Get a slice from index 2 to the end
print(nums[:2])     # Get a slice from the start to index 2 (exclusive)
print(nums[:])      # Get a slice of the whole list
print(nums[:-1])    # Slice indices can be negative
nums[2:4] = [8, 9] # Assign a new sublist to a slice
print(nums)
```

### Functions

Functions in programming languages allow you to encapsulate a piece of code such that you can use it in different places with different inputs without having to type in all the code again and again. Python functions are defined using the 'def' keyword. For example:

```python
def sign(x):
    if x > 0:
        return 'positive'
    elif x < 0:
        return 'negative'
    else:
        return 'zero'
```

```python
for x in [-1, 0, 1]:
    print(sign(x))
```

```python
#We will often define functions to take optional keyword arguments, like this:
def hello(name, loud=False):
    if loud:
        print(f'HELLO, {name.upper()}')
    else:
        print(f'Hello, {name}!')

hello('Bob')
hello('Fred', loud=True)
```

### Classes

A class takes the function concept one step further. It allows you to neatly encapsulate a collection of data and functions together. Let's look at an example:

```python
import numpy as np
class Circle:
    # The "constructor" - creates an instance of this class
    def __init__(self, radius, center_x, center_y):
        self.radius = radius
        self.center_x = center_x
        self.center_y = center_y

    def get_area(self):
        return np.pi * self.radius ** 2

    def get_circumference(self):
        return 2 * np.pi * self.radius

    def translate(self, move_x, move_y):
        self.center_x += move_x
        self.center_y += move_y

    def scale(self, percentage):
        self.radius += self.radius * percentage/100

    def print_class(self):
        print(f"Radius: {self.radius}")
        print(f"Center: ({self.center_x}, {self.center_y})")
```

Now we can define an instance of this class, get information from it, and perform operations on it.

```python
my_circle = Circle(radius=5, center_x=0, center_y=0) # or my_circle = Circle(5, 0, 0)
my_circle.print_class()
```
